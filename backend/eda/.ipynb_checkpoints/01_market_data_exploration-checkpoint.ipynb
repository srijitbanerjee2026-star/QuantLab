```python
import pandas as pd
import matplotlib.pyplot as plt
from app.core.database import SessionLocal
from app.services.market_data_service import get_market_data

# 1. Fetch AAPL data directly using your existing service layer
db = SessionLocal()
records = get_market_data(db, symbol="AAPL")
db.close()

# 2. Convert to Pandas DataFrame
df = pd.DataFrame([
    {
        "date": r.date,
        "open": float(r.open),
        "high": float(r.high),
        "low": float(r.low),
        "close": float(r.close),
        "adj_close": float(r.adj_close),
        "volume": r.volume
    }
    for r in records
])

df.set_index("date", inplace=True)
df.sort_index(inplace=True)

# 3. Calculate Simple Returns & 20-day Rolling Volatility
df["returns"] = df["adj_close"].pct_change()
df["volatility_20d"] = df["returns"].rolling(20).std() * (252 ** 0.5)  # Annualized

# 4. Display Summary & Plot
print(df.tail(10))
df[["adj_close"]].plot(title="AAPL Adjusted Close", figsize=(10, 4))
plt.show()
```
